```python
from Stream.simulatedStream import SimulatedStream
from Classes.classifierClass import Classifier
from Classes.filterClass import FilterClass
from Classes.featureClass import FeatureClass, FeatureType
from common import *
from scipy.io import loadmat
import numpy as np
import matplotlib.pyplot as plt
```

```python
mat = loadmat(resource_path("data.mat"),squeeze_me=True, struct_as_record=False)
```

```python
base_stream = SimulatedStream(stream_name="base", stream_type=4)
base_stream.start()
```

```python
base_filter = FilterClass(filter_name="alpha")
base_filter.filters = {'filter': ['bandpass'], 'order': [3], 'frequency': [[8, 12]]}
base_feature = FeatureClass(feature_type=FeatureType.ALPHA)
```

```python
base_classifier = Classifier(label0_datasets={'eyesOpen': mat['eyesOpen']}, label1_datasets={'eyesClosed': mat['eyesClosed']}, filters = [base_filter], features=[base_feature])
```

```python
base_classifier.train_model()
```
